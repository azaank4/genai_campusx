Building a custom multiplication tool using @tool decorator

```python
from langchain.tools import tool

@tool   # Using the tool decorator
def multiply(a : int, b : int) -> int:  # Mentioning that 'a' and 'b' are integers, and the output will also be an integer
    """This function can be used in case we need to multiply two numbers""" # Suggested to add docstring so model can understand the usecase of the given function
    return a * b

result = multiply.invoke({"a" : 5, "b" : 10})
print(result)
```

Custom tool using StructuredTool and Pydantic

```python
# StructuredTool seems deprecated and is no longer support by the latest versions of langchain
from langchain.tools.base import StructuredTool
from pydantic import BaseModel, Field

# Defining input constraints
class MultiplyInput(BaseModel):
    a : int = Field(required = True, description = "The first number to be multiplied")
    b : int = Field(required = True, description = "The second number to be multiplied")

# creating a function for multiplication
def multiply_func(a, b) -> int:
    return a * b

# Creating tool out of the function
multiply_tool = StructuredTool.from_function(
    function = multiply_func,   # Using the multiplication function
    name = "multiply",
    description = "Multiply two numbers",
    args_schema = MultiplyInput # Using the pydantic class we created
)

```
